# Завдання 1

```python
import numpy as np

arr = np.random.randint(0, 51, size=20)
print("Вихідний масив:")
print(arr)

threshold = int(input("\nВведіть порогове значення: "))

count_greater = np.sum(arr > threshold)
print(f"Кількість елементів, більших за {threshold}: {count_greater}")

max_val = np.max(arr)
max_idx = np.argmax(arr)
print(f"\nМаксимальне значення в масиві: {max_val}")
print(f"Позиція (індекс) першої появи: {max_idx}")

sorted_arr = np.sort(arr)[::-1]
print("\nВідсортований масив за спаданням:")
print(sorted_arr)
```

# Завдання 2

```python
import numpy as np

low = int(input("Введіть нижню межу: "))
high = int(input("Введіть верхню межу: "))

if low > high:
    low, high = high, low

matrix = np.random.randint(low, high + 1, size=(5, 5))
print("\nВихідна матриця 5x5:")
print(matrix)

main_diag = np.diag(matrix)
print("\nЕлементи головної діагоналі:")
print(main_diag)

diag_sum = np.sum(main_diag)
print(f"Сума елементів головної діагоналі: {diag_sum}")

modified_matrix = np.tril(matrix)
print("\nЗмінена матриця:")
print(modified_matrix)
```

# Завдання 3

```python
import numpy as np

start = int(input("Введіть початок діапазону: "))
end = int(input("Введіть кінець діапазону: "))

if (end - start + 1) != 30:
    seq = np.arange(start, start + 30)
else:
    seq = np.arange(start, end + 1)

print("\nСформована послідовність:")
print(seq)

matrix_6x5 = seq.reshape(6, 5)
print("\nМатриця 6x5:")
print(matrix_6x5)

row_sums = np.sum(matrix_6x5, axis=1)
print("\nСуми по рядках:")
print(row_sums)

col_maxs = np.max(matrix_6x5, axis=0)
print("\nМаксимуми по стовпчиках:")
print(col_maxs)
```
